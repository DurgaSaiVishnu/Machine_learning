```python
gojo_data = requests.get("https://www.amazon.in/s?k=gojo+saturo",headers=header,impersonate="chrome")
soup2 =  BeautifulSoup(gojo_data.text,"html.parser")
from bs4 import BeautifulSoup

soup = BeautifulSoup(gojo_data.text, "html.parser")

products = soup.find_all("div", {"data-component-type": "s-search-result"})

for product in products:
    title_element = product.find("h2")
    item_name = title_element.get_text(strip=True) if title_element else "Name not found"
    
    price_element = product.find("span", class_="a-price-whole")
    price = price_element.get_text(strip=True) if price_element else "Price not found"
    
    print(f"Product: {item_name}")
    print(f"Price: {price}")
    print("-" * 40)
```
